```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Load dataset
tips = sns.load_dataset("tips")
display(tips.head())
print("\n",tips.isnull().sum())
```

```python
# Scatter plot
plt.figure(figsize=(7,5))
sns.scatterplot(data=tips, x='total_bill', y='tip', hue='day',size='size')
plt.title("Tip vs Total Bill")
plt.xlabel("Total bill")
plt.ylabel("Tip")
plt.tight_layout()
plt.show()
```

```python
# Regression Plot
plt.figure(figsize=(7,4))
sns.regplot(data=tips, x='total_bill', y='tip')
plt.title("Trend : Total Bill vs Tip")
plt.xlabel("Total bill")
plt.ylabel("Tip")
plt.tight_layout()
plt.show()
```

```python
# Bar Plot
plt.figure(figsize=(7,5))
sns.barplot(data=tips, x='day', y='tip', errorbar=None)
plt.title("Average Tip by Day")
plt.xlabel("Day")
plt.ylabel("Tip")
plt.tight_layout()
plt.show()
```

```python
# Histogram
plt.figure(figsize=(7,4))
sns.histplot(data=tips, x='total_bill', bins=10, kde=True)
plt.title("Total bill distribution")
plt.xlabel("total bil")
plt.ylabel("count")
plt.tight_layout()
plt.show()
```

```python
# Box Plot
plt.figure(figsize=(7,5))
sns.boxplot(data=tips, x='day', y='total_bill')
plt.title("Total bill spread by day")
plt.xlabel("Day")
plt.ylabel("Total Bill")
plt.tight_layout()
plt.show()
```

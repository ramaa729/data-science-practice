```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.DataFrame({
    "hari" : ["senin", "selasa", "rabu", "kamis", "jumat"],
    "penjualan" : [12, 18, 19, 15, 16],
    "cabang" : ["A", "A", "A", "A", "A"],
})
print(df, "\n")
plt.figure(figsize=(6, 4))
sns.barplot(data=df, x="hari", y="penjualan")
plt.xlabel("Hari")
plt.ylabel("Jumlah terjual")
plt.title("Penjualan harian")
plt.show()
```

```python
df = pd.DataFrame({
    "hari": ["Sen", "Sel", "Rab", "Kam", "Sen", "Sel", "Rab", "Kam"],
    "penjualan": [12, 19, 15, 22, 10, 14, 18, 20],
    "cabang": ["Bandung", "Bandung", "Bandung", "Bandung", "Jakarta", "Jakarta", "Jakarta", "Jakarta"],
})
print(df, "\n")
plt.figure(figsize=(6, 4))
sns.lineplot(data=df, x="hari", y="penjualan", hue="cabang", marker="o")
plt.title("Penjualan harian per cabang")
plt.xlabel("hari")
plt.ylabel("jumlah terjual")
plt.show()
```

```python
df = pd.DataFrame({
    "kelas": ["A", "A", "A", "A", "A", "B", "B", "B", "B", "B"],
    "nilai": [70, 75, 80, 82, 95, 60, 68, 72, 74, 78],
})

print(df, "\n")

plt.figure(figsize=(6,4))
sns.boxplot(data=df, x="kelas", y="nilai")
plt.title("Sebaran nilai")
plt.xlabel("Kelas")
plt.ylabel("Nilai")
plt.show()
```

```python
df = pd.DataFrame({
    "jam_belajar": [1, 2, 3, 4, 5, 6],
    "nilai": [55, 62, 70, 75, 85, 92],
})

print(df, "\n")
plt.figure(figsize=(6, 4))
sns.scatterplot(data=df, x="jam_belajar", y="nilai")
plt.title("Hubungan jam belajar dengan nilai")
plt.xlabel("jam belajar")
plt.ylabel("nilai")
plt.show()
```

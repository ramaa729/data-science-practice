```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.DataFrame({
    "jam_belajar" : [2, 3, 4, 5, 6, 7, 2, 3, 5, 6, 8, 9],
    "nilai" : [58, 63, 68, 74, 79, 85, 61, 66, 77, 82, 89, 94],
    "jalur": [
        "Reguler", "Reguler", "Reguler", "Reguler", "Reguler", "Reguler",
        "Intensif", "Intensif", "Intensif", "Intensif", "Intensif", "Intensif",
    ],
})
print(df, "\n")

plt.figure(figsize=(6,4))
sns.lineplot(data=df, x="jam_belajar", y="nilai", hue="jalur", marker="o")
plt.xlabel("Jam belajar")
plt.ylabel("nilai")
plt.title("Performa jam belajar vs nilai ujian")
plt.show()
```

```python
plt.figure(figsize=(6,4))
sns.boxplot(data=df, x="jalur", y="nilai")
plt.xlabel("jalur")
plt.ylabel("nilai")
plt.title("sebaran nilai per jalur")
plt.show()
```

```python
df_fokus = pd.DataFrame({
    "jam_tidur": [5.0, 5.5, 6.0, 6.5, 7.0, 7.5, 8.0, 5.5, 6.5, 7.5],
    "skor_fokus": [52, 58, 63, 70, 76, 81, 85, 60, 72, 83],
    "kopi": ["Ya", "Ya", "Tidak", "Tidak", "Tidak", "Ya", "Tidak", "Ya", "Ya", "Tidak"],
})
print(df_fokus, "\n")

plt.figure(figsize=(6, 4))
sns.scatterplot(data=df_fokus, x="jam_tidur", y="skor_fokus", hue="kopi")
plt.title("hubungan waktu istirahat dengan skor fokus")
plt.xlabel("Jam tidur")
plt.ylabel("Skor fokus")
plt.show()
```
